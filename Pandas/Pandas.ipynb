```python
##boolean masking
data ={' planet': ['Mercury' , 'Vanus', 'Earth', 'Mars', 'Jupiter', 'Setrun', 'Uranus', 'Neptune'],
      ' radius_km': [2440,6052,6371,3390,69911,58232,25362,24622],
       'moons':[0,0,1,2,80,83,27,14]}
planets = pd.DataFrame(data)
planets
```

```python
mask = planets['moons'] < 20
mask
```

```python
planets[mask]
```

```python
planets[planets['moons']<20]
```

```python
planets
```

```python
moons_under_20 = planets[mask]
moons_under_20
```

```python
##logical operators
mask =( planets['moons'] <10) | (planets['moons'] > 50)
mask
```

```python
planets[mask]
```

```python
mask = (planets['moons'] > 20) & ~(planets['moons'] == 80) & ~(planets[' radius_km'] < 50000)
planets[mask]
```

```python
data ={' planet': ['Mercury' , 'Vanus', 'Earth', 'Mars', 'Jupiter', 'Setrun', 'Uranus', 'Neptune'],
      'radius_km': [2440,6052,6371,3390,69911,58232,25362,24622],
       'moons': [0,0,1,2,80,83,27,14],
       'type': ['terrestrail', 'terrestrail', 'terrestrail', 'terrestrail', 'gas giant', 'gas giant',
                 'ice giant', 'ice giant'],
       'rings': ['no', 'no', 'no', 'no', 'yes', 'yes', 'yes', 'yes'],
       'mean_temp_c': [167, 464, 15, -65, -110, -140, -195, -200],
       'megnatic_fields': ['yes', 'no', 'yes', 'no', 'yes', 'yes', 'yes', 'yes']
      }
planets = pd.DataFrame(data)
planets
```

```python
planets.groupby(['type'])
```

```python
planets.groupby('type').sum()
```

```python
planets.groupby('type')[['radius_km','moons','mean_temp_c']].agg(['mean','median'])

```

```python
planets.groupby('type')[['radius_km','moons','mean_temp_c']].agg(['mean','max'])
```

```python
def percentile_90(x):
     return x.quantile(0.9)
```

```python
planets.groupby('type')[['radius_km','moons','mean_temp_c']].agg(['mean',percentile_90])
```
